```python
import numpy as np
## euclidean distance calculation
# formula sqrt(sum((x + y)^2))
# x_arr and y_arr are np arrays, formula calculates every x_arr[i] with y_arr[i]

def euclidean_dist(x_arr, y_arr):
  s = np.sqrt(np.sum((x_arr - y_arr)**2))

  return s

## k nearest neighbours function:

def k_nearest_neighbors(x_train, y_train, test_a, k):

  distances = []
  label_dict = {}
  for i in range(len(x_train)):
    d = euclidean_dist(x_train[i], test_a)
    distances.append((d, y_train[i]))

    # label_dict[distances[i]] = y_train[i]

  distances.sort()
  sort_dist = distances[:k]

  # label_dict = dict(sorted(label_dict.items()))

  #   # array of labels


  # label_arr = []
  # for i in range(len(sort_dist)):
  #   label_arr.append(label_dict[i])
  #   # print(label_arr)

  return sort_dist



# k = 9
# print(k)


def knn_algoritm(k, t):
  avg_accur = 0
  acc_list = []
  ## Separate features and target. X -> data of the features, y -> classes of the data
  # x = df.drop("class", axis=1)
  # y = df["class"]

  for z in range(20):

    ## Split into training and testing sets 80% x_train, y_train to train the model, 20% x_test, y_test used to test the model
    x_train, x_test, y_train, y_test = train_test_split(
        x, y,
        test_size=0.2,
        shuffle = True

    )

    ## Scale the features
    scaler = StandardScaler()

    x_train = scaler.fit_transform(x_train)
    x_test = scaler.transform(x_test)

    y_train = y_train.to_numpy()
    y_test = y_test.to_numpy()



    count = 0

    if t == "train":

        for i in range(len(x_train)):
          test_a = x_train[i]

          q = k_nearest_neighbors(x_train, y_train, test_a, k)
          # for i in range(len(q)):
          #   print(q[i][0], "\t", q[i][1])

          ## doing the majority vote:

          labels = [l for d, l in q]

          prediction = Counter(labels).most_common(1)[0][0]


          ## calculating accuracy:

          if prediction == y_train[i]:
            count +=1

        accuracy = count / len(x_train)
        # print(z, "accuracy: ", accuracy)


    else:

        for i in range(len(x_test)):
          test_a = x_test[i]


          q = k_nearest_neighbors(x_train, y_train, test_a, k)
          # for i in range(len(q)):
          #   print(q[i][0], "\t", q[i][1])

          ## doing the majority vote:

          labels = [l for d, l in q]

          prediction = Counter(labels).most_common(1)[0][0]


          ## calculating accuracy:

          if prediction == y_test[i]:
            count +=1

        accuracy = count / len(x_test)
        # print(z, "accuracy: ", accuracy)

    avg_accur += accuracy
    acc_list.append(accuracy)
    # print(avg_accur)

  avg_ac = avg_accur / 20
  print("fo k", k, " average accuracy for the training set is = ", avg_ac)
  std_ac = np.std(acc_list)
  # print(std_ac)
  # print(acc_list)

  return avg_ac, std_ac
```
